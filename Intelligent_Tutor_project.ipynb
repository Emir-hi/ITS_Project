```python

import pandas as pd

# Load all four datasets
audio_df = pd.read_csv("D_audio.csv")
visual_df = pd.read_csv("D_visual.csv")
text_df = pd.read_csv("D_text.csv")
perf_df = pd.read_csv("D_perf.csv")

datasets = {
    "Audio": audio_df,
    "Visual": visual_df,
    "Text": text_df,
    "Performance": perf_df
}

# Check dimensions and student counts
for name, df in datasets.items():
    print(f"{name}:")
    print(f"  Shape: {df.shape}")
    print(f"  Students: {df['student_id'].nunique()}")
    print(f"  Sessions: {df['session_id'].nunique()}")
    print(f"  Missing values: {df.isnull().sum().sum()}")
    print("-" * 40)

# Check alignment of all interactions
keys = ["student_id", "session_id", "utterance_id"]

matched = audio_df[keys].merge(
    visual_df[keys], on=keys, validate="one_to_one"
).merge(
    text_df[keys], on=keys, validate="one_to_one"
).merge(
    perf_df[keys], on=keys, validate="one_to_one"
)

print("Matched interactions:", len(matched))
```

```python

from sklearn.model_selection import GroupShuffleSplit
from sklearn.preprocessing import StandardScaler
from sklearn.svm import SVC
from sklearn.pipeline import make_pipeline
from sklearn.metrics import (
    accuracy_score,
    classification_report,
    ConfusionMatrixDisplay
)
import matplotlib.pyplot as plt

# 1. Select input features
features = (
    [f"mfcc_{i}" for i in range(1, 14)]
    + ["pitch_hz", "energy_db", "speech_rate"]
)

X = audio_df[features]
y = audio_df["enthusiasm_label"]

# 2. Split by student to prevent data leakage
splitter = GroupShuffleSplit(
    n_splits=1,
    test_size=0.2,
    random_state=42
)

train_idx, test_idx = next(
    splitter.split(X, y, groups=audio_df["student_id"])
)

X_train = X.iloc[train_idx]
X_test = X.iloc[test_idx]
y_train = y.iloc[train_idx]
y_test = y.iloc[test_idx]

# 3. Build and train the SVM model
model = make_pipeline(
    StandardScaler(),
    SVC(
        kernel="rbf",
        class_weight="balanced",
        probability=True,
        random_state=42
    )
)

model.fit(X_train, y_train)

# 4. Make predictions
y_pred = model.predict(X_test)

# 5. Evaluate performance
print("Accuracy:", accuracy_score(y_test, y_pred))
print("\nClassification Report:")
print(classification_report(y_test, y_pred))

# 6. Visualize the confusion matrix
ConfusionMatrixDisplay.from_predictions(y_test, y_pred)
plt.title("SVM - Vocal Enthusiasm Classification")
plt.show()
```

```python

import pandas as pd
import numpy as np
from sklearn.feature_extraction.text import CountVectorizer
from sklearn.decomposition import LatentDirichletAllocation

# Load our audio dataset
audio_df = pd.read_csv("D_audio.csv")

# Select the spoken text
documents = audio_df["clean_tokens"].fillna("")

# Convert words into numerical counts
vectorizer = CountVectorizer(
    stop_words="english",
    min_df=3,
    max_df=0.95
)

X_counts = vectorizer.fit_transform(documents)

print("Document-term matrix:", X_counts.shape)

# Create the LDA model
lda_model = LatentDirichletAllocation(
    n_components=6,
    random_state=42,
    learning_method="batch",
    max_iter=25
)

# Train LDA and obtain topic probabilities
topic_probabilities = lda_model.fit_transform(X_counts)

# Display keywords for each discovered topic
words = vectorizer.get_feature_names_out()

for topic_id, topic in enumerate(lda_model.components_):
    top_indices = topic.argsort()[-10:][::-1]
    top_words = words[top_indices]

    print(f"\nTopic {topic_id}:")
    print(", ".join(top_words))
```

```python

import matplotlib.pyplot as plt

# Find the most probable topic for each response
dominant_topics = topic_probabilities.argmax(axis=1)

# Count the responses in each topic
topic_counts = pd.Series(dominant_topics).value_counts().sort_index()

plt.figure(figsize=(9, 5))
plt.bar(
    [f"Topic {i}" for i in topic_counts.index],
    topic_counts.values
)

plt.xlabel("Discovered Topic")
plt.ylabel("Number of Spoken Responses")
plt.title("LDA - Spoken Topic Distribution")
plt.show()
```

```python

# Create topic probability column names
topic_cols = [f"topic_{i}_prob" for i in range(6)]

# Create the output dataset
block2_output = audio_df[
    ["student_id", "session_id", "utterance_id"]
].copy()

# Add the probability distribution
block2_output[topic_cols] = topic_probabilities

# Add the dominant topic
block2_output["dominant_topic"] = (
    topic_probabilities.argmax(axis=1)
)

# Add the dominant topic's probability
block2_output["topic_confidence"] = (
    topic_probabilities.max(axis=1)
)

display(block2_output.head())

# Save the output
block2_output.to_csv(
    "B2_utterance_topics.csv",
    index=False
)
```

```python

# Average topic probabilities for each student
student_topics = (
    block2_output
    .groupby("student_id")[topic_cols]
    .mean()
    .reset_index()
)

# Find each student's dominant topic
scores = student_topics[topic_cols].to_numpy()

student_topics["favorite_topic_id"] = scores.argmax(axis=1)
student_topics["favorite_topic_score"] = scores.max(axis=1)

display(student_topics.head())

# Save the student-level output
student_topics.to_csv(
    "B2_student_topic_profile.csv",
    index=False
)
```

```python

import pandas as pd
import numpy as np
from sklearn.cluster import DBSCAN

# 1. Load the visual dataset
visual_df = pd.read_csv("D_visual.csv")

# 2. Select eye gaze coordinates
X = visual_df[["gaze_x", "gaze_y"]]

print("Input shape:", X.shape)

# 3. Create and train DBSCAN
dbscan = DBSCAN(
    eps=0.045,
    min_samples=20
)

# 4. Predict cluster assignments
visual_df["gaze_cluster"] = dbscan.fit_predict(X)

# 5. Examine discovered clusters
n_clusters = visual_df.loc[
    visual_df["gaze_cluster"] != -1,
    "gaze_cluster"
].nunique()

n_noise = (visual_df["gaze_cluster"] == -1).sum()

print("Discovered clusters:", n_clusters)
print("Noise points:", n_noise)

print("\nCluster distribution:")
print(
    visual_df["gaze_cluster"]
    .value_counts()
    .sort_index()
)
```

```python

import matplotlib.pyplot as plt

plt.figure(figsize=(10, 7))

# Draw each discovered cluster separately
for cluster_id, group in visual_df.groupby("gaze_cluster"):

    label = (
        "Noise"
        if cluster_id == -1
        else f"Cluster {cluster_id}"
    )

    plt.scatter(
        group["gaze_x"],
        group["gaze_y"],
        label=label,
        alpha=0.55,
        s=12
    )

plt.xlabel("Horizontal Gaze Position (X)")
plt.ylabel("Vertical Gaze Position (Y)")
plt.title("DBSCAN - Eye Gaze Clustering")

plt.xlim(0, 1)
plt.ylim(0, 1)
plt.gca().invert_yaxis()

plt.legend()
plt.grid(alpha=0.2)
plt.show()
```

```python

# Use screen-region metadata to interpret clusters
cluster_topic_map = (
    visual_df[visual_df["gaze_cluster"] != -1]
    .groupby("gaze_cluster")["aoi_topic_tag"]
    .agg(lambda x: x.value_counts().idxmax())
    .to_dict()
)

print("Cluster-to-topic mapping:")
for cluster_id, topic in sorted(cluster_topic_map.items()):
    print(f"Cluster {cluster_id}: {topic}")

# Add topic names to our results
visual_df["attended_topic"] = (
    visual_df["gaze_cluster"]
    .map(cluster_topic_map)
    .fillna("Unassigned")
)

display(
    visual_df[
        ["student_id", "gaze_x", "gaze_y",
         "fixation_ms", "gaze_cluster", "attended_topic"]
    ].head()
)
```

```python

# Define the topics
topics = [
    "Space", "Sports", "Technology",
    "Music", "Nature", "History"
]

# Keep gaze points assigned to topic clusters
assigned = visual_df[
    visual_df["gaze_cluster"] != -1
].copy()

# Sum fixation time for each student and topic
attention = (
    assigned
    .groupby(["student_id", "attended_topic"])
    ["fixation_ms"]
    .sum()
    .unstack(fill_value=0)
)

# Include all students and all six topics
attention = attention.reindex(
    index=sorted(visual_df["student_id"].unique()),
    columns=topics,
    fill_value=0
)

# Convert fixation durations into attention shares
totals = attention.sum(axis=1)

attention_scores = (
    attention.div(totals.replace(0, np.nan), axis=0)
    .fillna(0)
)

# Create student profile dataset
student_gaze_profile = attention_scores.add_prefix(
    "gaze_share_"
).reset_index()

# Find most attended topic
student_gaze_profile["top_visual_topic"] = np.where(
    totals.to_numpy() > 0,
    attention_scores.idxmax(axis=1).to_numpy(),
    "Unknown"
)

display(student_gaze_profile.head())
```

```python

# Output 1: Observation-level predictions
block3_output = visual_df[
    [
        "student_id", "session_id",
        "utterance_id", "frame_id",
        "gaze_cluster", "attended_topic",
        "fixation_ms"
    ]
]

block3_output.to_csv(
    "B3_gaze_clusters.csv",
    index=False
)

# Output 2: Student-level visual profile
student_gaze_profile.to_csv(
    "B3_student_gaze_profile.csv",
    index=False
)

print("Block 3 output files saved successfully.")
print("Observations:", block3_output.shape)
print("Student profiles:", student_gaze_profile.shape)
```

```python

import pandas as pd
import numpy as np
import json

# Load visual dataset
visual_df = pd.read_csv("D_visual.csv")

# Convert 68 facial landmarks from JSON into numbers
landmarks = np.array(
    visual_df["landmarks_68"].apply(json.loads).tolist(),
    dtype=float
)

print("Landmarks shape:", landmarks.shape)

# Each landmark has an X and Y coordinate
landmark_columns = [
    f"lm_{i:02d}_{axis}"
    for i in range(1, 69)
    for axis in ["x", "y"]
]

landmark_df = pd.DataFrame(
    landmarks.reshape(len(visual_df), 136),
    columns=landmark_columns
)

# Select head-pose and facial-expression features
basic_features = [
    "head_yaw", "head_pitch", "head_roll",
    "au_01", "au_02", "au_03",
    "au_04", "au_05", "au_06"
]

# Combine all numerical inputs
X = pd.concat(
    [
        visual_df[basic_features].reset_index(drop=True),
        landmark_df
    ],
    axis=1
)

# Target labels
y = visual_df["engagement_label"]

print("Input shape:", X.shape)
print("Target shape:", y.shape)
print("\nEngagement distribution:")
print(y.value_counts())
```

```python

from sklearn.ensemble import RandomForestClassifier
from sklearn.model_selection import GroupShuffleSplit
from sklearn.metrics import (
    accuracy_score,
    classification_report,
    ConfusionMatrixDisplay
)
import matplotlib.pyplot as plt

# Split students into training and testing groups
splitter = GroupShuffleSplit(
    n_splits=1,
    test_size=0.2,
    random_state=42
)

train_idx, test_idx = next(
    splitter.split(
        X, y,
        groups=visual_df["student_id"]
    )
)

X_train = X.iloc[train_idx]
X_test = X.iloc[test_idx]
y_train = y.iloc[train_idx]
y_test = y.iloc[test_idx]

# Create Random Forest model
rf_model = RandomForestClassifier(
    n_estimators=200,
    min_samples_leaf=3,
    class_weight="balanced_subsample",
    random_state=42,
    n_jobs=-1
)

# Train the model
rf_model.fit(X_train, y_train)

# Predict engagement
y_pred = rf_model.predict(X_test)

# Evaluate
print("Accuracy:", accuracy_score(y_test, y_pred))

print("\nClassification Report:")
print(classification_report(y_test, y_pred))

# Display confusion matrix
labels = ["Intrigued", "Focused", "Bored"]

ConfusionMatrixDisplay.from_predictions(
    y_test,
    y_pred,
    labels=labels,
    cmap="Blues",
    values_format="d"
)

plt.title("Random Forest - Facial Engagement")
plt.show()
```

```python

# Calculate feature importance
importances = pd.Series(
    rf_model.feature_importances_,
    index=X.columns
)

# Select the top 12
top_features = importances.nlargest(12)

# Visualization
plt.figure(figsize=(10, 6))
top_features.sort_values().plot.barh()

plt.xlabel("Feature Importance")
plt.title("Random Forest - Top 12 Facial Features")
plt.tight_layout()
plt.show()

print(top_features)
```

```python

from sklearn.model_selection import GroupKFold, cross_val_predict

# 5-fold student-grouped cross-validation
cv = GroupKFold(n_splits=5)

# Out-of-fold class probabilities
oof_probs = cross_val_predict(
    RandomForestClassifier(
        n_estimators=200,
        min_samples_leaf=3,
        class_weight="balanced_subsample",
        random_state=42,
        n_jobs=-1
    ),
    X,
    y,
    groups=visual_df["student_id"],
    cv=cv,
    method="predict_proba"
)

# Class order is alphabetical
classes = np.sort(y.unique())
prob_cols = [f"prob_{c.lower()}" for c in classes]

# Create observation-level output
block4_output = visual_df[
    ["student_id", "session_id", "utterance_id", "frame_id"]
].copy()

block4_output[prob_cols] = oof_probs

block4_output["predicted_engagement"] = (
    classes[oof_probs.argmax(axis=1)]
)

block4_output["prediction_confidence"] = (
    oof_probs.max(axis=1)
)

# Save observation-level predictions
block4_output.to_csv(
    "B4_engagement_predictions.csv",
    index=False
)

# Average class probabilities for each student
student_engagement = (
    block4_output.groupby("student_id")[prob_cols]
    .mean()
    .reset_index()
)

# Identify each student's dominant engagement category
student_engagement["dominant_engagement"] = (
    classes[
        student_engagement[prob_cols].to_numpy().argmax(axis=1)
    ]
)

# Save student-level profile
student_engagement.to_csv(
    "B4_student_engagement_profile.csv",
    index=False
)

print("Observation output:", block4_output.shape)
print("Student profiles:", student_engagement.shape)

display(student_engagement.head())
```

```python

import pandas as pd
import numpy as np

# Load text dataset
text_df = pd.read_csv("D_text.csv")

# Select input text
X = text_df["clean_tokens"].fillna("").astype(str)

# Two supervised targets
y_topic = text_df["topic_id"]
y_curiosity = text_df["curiosity_polarity"]

print("Dataset shape:", text_df.shape)

print("\nTopic distribution:")
print(y_topic.value_counts())

print("\nCuriosity distribution:")
print(y_curiosity.value_counts())

display(text_df[
    ["raw_text", "clean_tokens",
     "topic_id", "curiosity_polarity"]
].head())
```

```python

from sklearn.model_selection import GroupShuffleSplit
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.naive_bayes import MultinomialNB
from sklearn.pipeline import make_pipeline

# Split by student
splitter = GroupShuffleSplit(
    n_splits=1,
    test_size=0.2,
    random_state=42
)

train_idx, test_idx = next(
    splitter.split(
        X,
        groups=text_df["student_id"]
    )
)

X_train = X.iloc[train_idx]
X_test = X.iloc[test_idx]

# Function for creating a Naive Bayes pipeline
def create_nb_model():
    return make_pipeline(
        TfidfVectorizer(
            ngram_range=(1, 2),
            min_df=2,
            sublinear_tf=True
        ),
        MultinomialNB(alpha=0.5)
    )

# Model 1: Topic classification
topic_model = create_nb_model()

topic_model.fit(
    X_train,
    y_topic.iloc[train_idx]
)

# Model 2: Curiosity classification
curiosity_model = create_nb_model()

curiosity_model.fit(
    X_train,
    y_curiosity.iloc[train_idx]
)

print("Both Naive Bayes models trained successfully!")
```

```python

from sklearn.metrics import (
    accuracy_score,
    classification_report,
    ConfusionMatrixDisplay
)
import matplotlib.pyplot as plt

# Topic predictions
topic_pred = topic_model.predict(X_test)

# Curiosity predictions
curiosity_pred = curiosity_model.predict(X_test)

# Evaluate topic model
print("=== TOPIC CLASSIFICATION ===")

print("Accuracy:", accuracy_score(
    y_topic.iloc[test_idx], topic_pred
))

print(classification_report(
    y_topic.iloc[test_idx], topic_pred
))

# Evaluate curiosity model
print("\n=== CURIOSITY CLASSIFICATION ===")

print("Accuracy:", accuracy_score(
    y_curiosity.iloc[test_idx], curiosity_pred
))

print(classification_report(
    y_curiosity.iloc[test_idx], curiosity_pred
))

# Topic confusion matrix
ConfusionMatrixDisplay.from_predictions(
    y_topic.iloc[test_idx],
    topic_pred,
    cmap="Blues",
    xticks_rotation=45
)

plt.title("Naive Bayes - Topic Classification")
plt.tight_layout()
plt.show()

# Curiosity confusion matrix
ConfusionMatrixDisplay.from_predictions(
    y_curiosity.iloc[test_idx],
    curiosity_pred,
    cmap="Blues"
)

plt.title("Naive Bayes - Curiosity Classification")
plt.tight_layout()
plt.show()
```

```python

from sklearn.model_selection import GroupKFold, cross_val_predict

cv = GroupKFold(n_splits=5)

# Out-of-fold topic probabilities
topic_probs = cross_val_predict(
    create_nb_model(),
    X,
    y_topic,
    cv=cv,
    groups=text_df["student_id"],
    method="predict_proba"
)

# Out-of-fold curiosity probabilities
curiosity_probs = cross_val_predict(
    create_nb_model(),
    X,
    y_curiosity,
    cv=cv,
    groups=text_df["student_id"],
    method="predict_proba"
)

# Get target class names in alphabetical order
topic_classes = np.sort(y_topic.unique())
curiosity_classes = np.sort(y_curiosity.unique())

# Create observation-level output
block5_output = text_df[
    ["student_id", "session_id", "utterance_id", "msg_id"]
].copy()

# Add topic probabilities
for i, topic in enumerate(topic_classes):
    block5_output[f"topic_prob_{topic.lower()}"] = (
        topic_probs[:, i]
    )

# Add curiosity probabilities
for i, label in enumerate(curiosity_classes):
    block5_output[f"curiosity_prob_{label.lower()}"] = (
        curiosity_probs[:, i]
    )

# Add predicted classes
block5_output["predicted_topic"] = (
    topic_classes[topic_probs.argmax(axis=1)]
)

block5_output["predicted_curiosity"] = (
    curiosity_classes[curiosity_probs.argmax(axis=1)]
)

# Save message-level predictions
block5_output.to_csv(
    "B5_text_predictions.csv",
    index=False
)

# Combine topic and curiosity into interest signals
curiosity_weight = (
    block5_output["curiosity_prob_curious"]
    + 0.5 * block5_output["curiosity_prob_neutral"]
)

score_cols = []

for topic in topic_classes:
    name = topic.lower()
    col = f"text_interest_{name}"

    block5_output[col] = (
        block5_output[f"topic_prob_{name}"]
        * curiosity_weight
    )

    score_cols.append(col)

# Average scores for each student
student_text_profile = (
    block5_output
    .groupby("student_id")[score_cols]
    .mean()
    .reset_index()
)

student_text_profile["top_text_topic"] = (
    student_text_profile[score_cols]
    .idxmax(axis=1)
    .str.replace("text_interest_", "", regex=False)
    .str.title()
)

# Average predicted curiosity probability
avg_curiosity = (
    block5_output
    .groupby("student_id")["curiosity_prob_curious"]
    .mean()
)

student_text_profile["avg_curiosity_probability"] = (
    student_text_profile["student_id"]
    .map(avg_curiosity)
)

# Save student-level profiles
student_text_profile.to_csv(
    "B5_student_text_profile.csv",
    index=False
)

print("Message output:", block5_output[
    ["student_id", "session_id", "utterance_id", "msg_id"]
    + [c for c in block5_output.columns
       if c.startswith("topic_prob_")
       or c.startswith("curiosity_prob_")]
    + ["predicted_topic", "predicted_curiosity"]
].shape)

print("Student profiles:", student_text_profile.shape)

display(student_text_profile.head())
```

```python

import pandas as pd
import numpy as np

# Load performance dataset
perf_df = pd.read_csv("D_perf.csv")

# Numerical input features
numeric_features = [
    "difficulty_level",
    "attempt_num",
    "prior_skill_attempts",
    "prior_skill_accuracy",
    "prior_skill_last_correct",
    "prior_overall_attempts",
    "prior_overall_accuracy",
    "prior_session_accuracy",
    "prior_avg_response_time_ms",
    "prior_avg_hints"
]

# Categorical input features
categorical_features = [
    "skill_id",
    "exercise_type",
    "topic_id"
]

# Separate inputs and target
X_perf = perf_df[numeric_features + categorical_features]
y_perf = perf_df["is_correct"]

print("Dataset shape:", perf_df.shape)
print("Input shape:", X_perf.shape)

print("\nAnswer distribution:")
print(y_perf.value_counts())

display(perf_df.head())
```

```python

from sklearn.model_selection import GroupShuffleSplit
from sklearn.preprocessing import StandardScaler, OneHotEncoder
from sklearn.compose import ColumnTransformer
from sklearn.pipeline import Pipeline
from sklearn.linear_model import LogisticRegression

# Split by student
splitter = GroupShuffleSplit(
    n_splits=1,
    test_size=0.2,
    random_state=42
)

train_idx, test_idx = next(
    splitter.split(
        X_perf,
        y_perf,
        groups=perf_df["student_id"]
    )
)

X_train = X_perf.iloc[train_idx]
X_test = X_perf.iloc[test_idx]

y_train = y_perf.iloc[train_idx]
y_test = y_perf.iloc[test_idx]

# Prepare numerical and categorical features
preprocessor = ColumnTransformer([
    ("num", StandardScaler(), numeric_features),
    ("cat", OneHotEncoder(handle_unknown="ignore"),
     categorical_features)
])

# Create Logistic Regression pipeline
kt_model = Pipeline([
    ("preprocess", preprocessor),
    ("model", LogisticRegression(
        max_iter=1000,
        random_state=42
    ))
])

# Train
kt_model.fit(X_train, y_train)

print("Knowledge Tracing model trained successfully!")
```

```python

from sklearn.metrics import (
    accuracy_score,
    classification_report,
    confusion_matrix,
    ConfusionMatrixDisplay,
    roc_auc_score
)
import matplotlib.pyplot as plt

# Make predictions
y_pred = kt_model.predict(X_test)

# Probability of a correct answer
y_prob = kt_model.predict_proba(X_test)[:, 1]

# Evaluate
print("Accuracy:", accuracy_score(y_test, y_pred))
print("ROC-AUC:", roc_auc_score(y_test, y_prob))

print("\nClassification Report:")
print(classification_report(y_test, y_pred))

# Confusion matrix
ConfusionMatrixDisplay.from_predictions(
    y_test,
    y_pred,
    display_labels=["Incorrect", "Correct"],
    cmap="Blues"
)

plt.title("Logistic Regression - Knowledge Tracing")
plt.show()
```

```python

# Select an unseen exercise from the test set
sample = X_test.iloc[[0]]

# Predict probability of success
probability = kt_model.predict_proba(sample)[0, 1]

print("Skill:", sample["skill_id"].iloc[0])
print("Difficulty:", sample["difficulty_level"].iloc[0])

print(f"Probability of correct answer: {probability:.2%}")

if probability < 0.40:
    print("Recommendation: Easier exercise with guidance")
elif probability < 0.75:
    print("Recommendation: Moderate exercise with optional hints")
else:
    print("Recommendation: More challenging exercise")
```

```python

from sklearn.model_selection import GroupKFold, cross_val_predict

# Five-fold cross-validation grouped by student
cv = GroupKFold(n_splits=5)

# Generate out-of-fold probabilities
oof_probs = cross_val_predict(
    kt_model,
    X_perf,
    y_perf,
    groups=perf_df["student_id"],
    cv=cv,
    method="predict_proba"
)[:, 1]

# Create exercise-level output
block7_output = perf_df[
    [
        "student_id",
        "session_id",
        "utterance_id",
        "attempt_id",
        "skill_id"
    ]
].copy()

block7_output["p_correct"] = oof_probs

block7_output["predicted_correct"] = (
    oof_probs >= 0.5
).astype(int)

# Save exercise predictions
block7_output.to_csv(
    "B7_attempt_predictions.csv",
    index=False
)

# Create student-by-skill probability matrix
skills = sorted(perf_df["skill_id"].unique())
students = sorted(perf_df["student_id"].unique())

skill_scores = block7_output.pivot_table(
    index="student_id",
    columns="skill_id",
    values="p_correct",
    aggfunc="mean"
).reindex(index=students, columns=skills)

# Flag skills without observations
observed = skill_scores.notna().astype(int)

# Use a neutral placeholder for unobserved skills
skill_scores = skill_scores.fillna(0.5)

skill_scores.columns = [
    f"mastery_{s}" for s in skill_scores.columns
]

observed.columns = [
    f"observed_{s}" for s in observed.columns
]

student_skill_profile = pd.concat(
    [skill_scores, observed],
    axis=1
).reset_index()

# Save profile
student_skill_profile.to_csv(
    "B7_student_skill_profile.csv",
    index=False
)

print("Attempt predictions:", block7_output.shape)
print("Student skill profiles:", student_skill_profile.shape)

display(student_skill_profile.head())
```

```python

import pandas as pd
import numpy as np

# Load performance dataset
perf_df = pd.read_csv("D_perf.csv")

# Aggregate exercise data for each student
behavior_df = (
    perf_df.groupby("student_id")
    .agg(
        n_attempts=("attempt_id", "count"),
        accuracy=("is_correct", "mean"),
        avg_response_time_s=("response_time_ms",
                             lambda x: x.mean() / 1000),
        avg_hints=("hint_count", "mean"),
        retry_rate=("attempt_num",
                    lambda x: (x > 1).mean()),
        avg_difficulty=("difficulty_level", "mean")
    )
    .reset_index()
)

print("Behavior dataset shape:", behavior_df.shape)

display(behavior_df.head())
```

```python

from sklearn.cluster import KMeans
from sklearn.preprocessing import StandardScaler
from sklearn.metrics import silhouette_score

# Select clustering features
behavior_features = [
    "accuracy",
    "avg_response_time_s",
    "avg_hints",
    "retry_rate"
]

X_behavior = behavior_df[behavior_features]

# Standardize numerical features
scaler = StandardScaler()
X_scaled = scaler.fit_transform(X_behavior)

# Create K-Means model
kmeans = KMeans(
    n_clusters=3,
    random_state=42,
    n_init=10
)

# Train the model
behavior_df["behavior_cluster"] = (
    kmeans.fit_predict(X_scaled)
)

# Evaluate cluster separation
silhouette = silhouette_score(
    X_scaled,
    behavior_df["behavior_cluster"]
)

print("Silhouette Score:", round(silhouette, 3))

print("\nStudents per cluster:")
print(behavior_df["behavior_cluster"].value_counts())

display(behavior_df.head())
```

```python

import matplotlib.pyplot as plt
from sklearn.decomposition import PCA

# Reduce four features to two dimensions
pca = PCA(n_components=2)

X_pca = pca.fit_transform(X_scaled)

plt.figure(figsize=(10, 7))

for cluster_id in sorted(
    behavior_df["behavior_cluster"].unique()
):
    mask = behavior_df["behavior_cluster"] == cluster_id

    plt.scatter(
        X_pca[mask, 0],
        X_pca[mask, 1],
        label=f"Cluster {cluster_id}",
        alpha=0.7,
        s=35
    )

plt.xlabel("Principal Component 1")
plt.ylabel("Principal Component 2")
plt.title("K-Means - Student Learning Behaviors")
plt.legend()
plt.grid(alpha=0.2)
plt.show()
```

```python

# Names based on the observed cluster characteristics
cluster_names = {
    0: "Needs More Support",
    1: "Efficient Independent",
    2: "Persistent With Retries"
}

# Add descriptive cluster labels
behavior_df["behavior_type"] = (
    behavior_df["behavior_cluster"].map(cluster_names)
)

# Save output
behavior_df.to_csv(
    "B8_student_behavior_clusters.csv",
    index=False
)

print("Block 8 output saved successfully!")
print("Output shape:", behavior_df.shape)

display(behavior_df.head())
```

```python

import pandas as pd
import numpy as np

from sklearn.cluster import KMeans
from sklearn.preprocessing import StandardScaler
from sklearn.metrics import silhouette_score

# Load our fifth dataset
student_df = pd.read_csv("D_student.csv")

# Select clustering features
features = ["interest_score", "mastery_score"]

X = student_df[features]

# Standardize features
scaler = StandardScaler()
X_scaled = scaler.fit_transform(X)

# Train K-Means with four clusters
kmeans = KMeans(
    n_clusters=4,
    random_state=42,
    n_init=20
)

student_df["cluster_id"] = (
    kmeans.fit_predict(X_scaled)
)

# Evaluate clusters
silhouette = silhouette_score(
    X_scaled,
    student_df["cluster_id"]
)

print("Silhouette Score:", round(silhouette, 3))

# Convert cluster centers back to original scale
centers = scaler.inverse_transform(
    kmeans.cluster_centers_
)

cluster_centers = pd.DataFrame(
    centers,
    columns=features
)

cluster_centers["cluster_id"] = range(4)

display(cluster_centers)
```

```python

from scipy.optimize import linear_sum_assignment

# Scale cluster centers between 0 and 1
center_min = centers.min(axis=0)
center_range = np.ptp(centers, axis=0)

center_range = np.where(
    center_range == 0, 1, center_range
)

normalized_centers = (
    (centers - center_min) / center_range
)

# Ideal category patterns:
# A: High interest, Low mastery
# B: High interest, High mastery
# C: Low interest, High mastery
# D: Low interest, Low mastery
category_patterns = np.array([
    [1, 0],
    [1, 1],
    [0, 1],
    [0, 0]
])

category_names = ["A", "B", "C", "D"]

# Calculate distances to category patterns
cost = (
    (
        normalized_centers[:, None, :]
        - category_patterns[None, :, :]
    ) ** 2
).sum(axis=2)

# Find unique category assignments
rows, cols = linear_sum_assignment(cost)

category_map = {
    int(row): category_names[int(col)]
    for row, col in zip(rows, cols)
}

student_df["category"] = (
    student_df["cluster_id"].map(category_map)
)

print("Cluster-to-category mapping:")
print(category_map)

print("\nStudents per category:")
print(student_df["category"].value_counts().sort_index())

print("\nCategory averages:")
display(
    student_df.groupby("category")[
        ["interest_score", "mastery_score", "struggle_score"]
    ].mean().round(3)
)
```

```python

import matplotlib.pyplot as plt

plt.figure(figsize=(10, 7))

for category in ["A", "B", "C", "D"]:

    group = student_df[
        student_df["category"] == category
    ]

    plt.scatter(
        group["interest_score"],
        group["mastery_score"],
        label=f"Category {category}",
        alpha=0.65,
        s=35
    )

plt.xlabel("Multimodal Interest Score")
plt.ylabel("Predicted Skill Performance")
plt.title("K-Means - Student Categorization")

plt.legend()
plt.grid(alpha=0.2)
plt.tight_layout()
plt.show()
```

```python

def recommend_lesson(student):

    category = student["category"]
    topic = student["top_topics"].split(" | ")[0]
    skill = student["weak_skills"]

    if category == "A":
        return (
            f"Practice {skill} using {topic} examples "
            "with step-by-step hints."
        )

    elif category == "B":
        return (
            f"Try advanced {skill} exercises "
            f"related to {topic}."
        )

    elif category == "C":
        return (
            "Introduce a fresh discussion topic "
            "with challenging English activities."
        )

    else:
        return (
            f"Start with easy {skill} exercises "
            f"using fun {topic} examples."
        )


# Generate a recommendation for every student
student_df["recommendation"] = (
    student_df.apply(recommend_lesson, axis=1)
)

# Display example students
display(
    student_df[
        [
            "student_id",
            "category",
            "top_topics",
            "weak_skills",
            "recommendation"
        ]
    ].head(10)
)
```

Cell 2 — Configure the policy
- Starting difficulty: A / D = Easy, C = Medium, B = Hard.
- After each answer: correct +0.012, incorrect with hint +0.006, incorrect without hint +0.002 to the targeted skill state.
- After each week: accuracy ≥ 80% = level up, accuracy ≤ 40% = level down, otherwise unchanged (levels 1–3).
- Selection: lean toward high-interest topics and weaker English skills; avoid repeating questions until that difficulty's unused questions are exhausted.
These are explicit simulation assumptions, not trained model parameters.

```python
WEEKS = 8
QUESTIONS_PER_WEEK = 5
RANDOM_SEED = 42
rng = np.random.default_rng(RANDOM_SEED)

SKILLS = ["vocabulary", "past_tense", "present_tense",
          "prepositions", "reading_comprehension"]
TOPICS = ["Space", "Sports", "Technology", "Music", "Nature", "History"]
START_LEVEL = {"A": 1, "B": 3, "C": 2, "D": 1}
HINT_RATES = {"A": 0.55, "B": 0.10, "C": 0.25, "D": 0.45}

# Prebuild exercise lists so repeated selection is fast.
questions_by_level = {
    level: exercises.loc[exercises["difficulty"] == level].to_dict("records")
    for level in (1, 2, 3)
}

def new_student_state(student):
    """Create the changing state; Stage 3 categories and interests stay as baseline."""
    return {
        "difficulty": START_LEVEL[str(student["category"])],
        "skill_state": {
            skill: float(np.clip(student[f"mastery_{skill}"], 0.05, 0.95))
            for skill in SKILLS
        },
        "used_ids": set()
    }

def choose_next_exercise(student, state, rng):
    """The adaptive tutoring policy: dynamic difficulty + weakest-skill bias."""
    level = state["difficulty"]
    pool = questions_by_level[level]
    unseen = [q for q in pool if q["exercise_id"] not in state["used_ids"]]
    repeating = not bool(unseen)
    pool = unseen if unseen else pool

    ranked = sorted(TOPICS, key=lambda t: -float(student[f"interest_{t.lower()}"]))
    if student["category"] == "C":  # Encourage an alternative to the top two topics.
        ranked = ranked[2:] + ranked[:2]
    topic_weight = {topic: 3.5 - 0.5*i for i, topic in enumerate(ranked)}

    weights = np.array([
        topic_weight[q["topic"]] *
        (0.5 + 2.0 * (1 - state["skill_state"][q["skill"]]))
        for q in pool
    ], dtype=float)
    weights /= weights.sum()
    question = pool[int(rng.choice(len(pool), p=weights))]
    return question, repeating

def update_skill_after_feedback(state, skill, correct, used_hint):
    """Revised rule: correct responses give the largest practice gain."""
    gain = 0.012 if correct else (0.006 if used_hint else 0.002)
    before = state["skill_state"][skill]
    state["skill_state"][skill] = min(0.95, before + gain)
    return before, state["skill_state"][skill], gain

def adapt_difficulty(state, weekly_accuracy):
    """Use *last week's observed simulated responses* for the next week."""
    if weekly_accuracy >= 0.80:
        state["difficulty"] = min(3, state["difficulty"] + 1)
    elif weekly_accuracy <= 0.40:
        state["difficulty"] = max(1, state["difficulty"] - 1)
```

Cell 3 — Simulate student responses and run the full feedback loop

```python
def wrong_answer_for(exercise, rng):
    """Return a plausible wrong option/form, guaranteed to differ from the key."""
    answer = str(exercise["correct_answer"]).strip()
    if exercise["exercise_type"] == "multiple_choice":
        return str(rng.choice([x for x in "ABC" if x != answer.upper()]))
    if exercise["skill"] == "prepositions":
        return str(rng.choice([x for x in ["in", "on", "at", "to", "into", "under"]
                               if x != answer.lower()]))
    if answer.startswith(("is ", "has ")):
        proposed = answer.split(" ", 1)[1]
    elif answer.endswith("ed"):
        proposed = answer[:-2] + "ing"
    elif answer.endswith("s"):
        proposed = answer[:-1]
    else:
        proposed = "I don't know"
    return proposed if proposed.lower() != answer.lower() else "I don't know"

attempt_rows = []
weekly_rows = []
final_rows = []
state_by_student = {}

for _, student in students.iterrows():
    student_id = str(student["student_id"])
    category = str(student["category"])
    struggle = float(student["struggle_score"])
    state = new_student_state(student)

    for week in range(1, WEEKS + 1):
        this_week = []
        level_this_week = state["difficulty"]

        for question_no in range(1, QUESTIONS_PER_WEEK + 1):
            question, repeated = choose_next_exercise(student, state, rng)
            state["used_ids"].add(question["exercise_id"])
            skill = question["skill"]
            level = int(question["difficulty"])
            ability_before = state["skill_state"][skill]

            used_hint = int(rng.random() < min(0.90, HINT_RATES[category] + 0.10*struggle))
            probability = float(np.clip(
                ability_before - 0.11 * (level - 1)
                - 0.08 * (struggle - 0.4) + 0.07 * used_hint,
                0.08, 0.95
            ))
            correct = int(rng.random() < probability)
            answer = (str(question["correct_answer"]) if correct
                      else wrong_answer_for(question, rng))

            _, ability_after, gain = update_skill_after_feedback(
                state, skill, correct, used_hint
            )
            this_week.append(correct)
            attempt_rows.append({
                "student_id": student_id, "category": category,
                "week": week, "question_no": question_no,
                "exercise_id": question["exercise_id"],
                "topic": question["topic"], "skill": skill,
                "difficulty": level, "question": question["question"],
                "student_answer": answer,
                "correct_answer": str(question["correct_answer"]),
                "used_hint": used_hint, "is_correct": correct,
                "p_correct_simulated": round(probability, 4),
                "simulated_skill_before": round(ability_before, 4),
                "simulated_skill_after": round(ability_after, 4),
                "learning_gain": gain, "repeat_question": repeated
            })

        accuracy = float(np.mean(this_week))
        adapt_difficulty(state, accuracy)
        weekly_rows.append({
            "student_id": student_id, "category": category, "week": week,
            "questions_answered": QUESTIONS_PER_WEEK,
            "correct_answers": sum(this_week),
            "weekly_accuracy": round(accuracy, 4),
            "difficulty_this_week": level_this_week,
            "difficulty_next_week": state["difficulty"],
            "simulated_avg_skill_state": round(
                float(np.mean(list(state["skill_state"].values()))), 4
            ),
            "level_change": state["difficulty"] - level_this_week
        })

    state_by_student[student_id] = state
    final_rows.append({
        "student_id": student_id, "category": category,
        "final_difficulty": state["difficulty"],
        **{f"simulated_final_{skill}": state["skill_state"][skill]
           for skill in SKILLS}
    })

attempts = pd.DataFrame(attempt_rows)
weekly = pd.DataFrame(weekly_rows)
final_profiles = pd.DataFrame(final_rows)

assert len(attempts) == len(students) * WEEKS * QUESTIONS_PER_WEEK
assert len(weekly) == len(students) * WEEKS
assert attempts["difficulty"].between(1, 3).all()
assert attempts["is_correct"].isin([0, 1]).all()
assert all(attempts["student_answer"].str.strip().ne("") )
assert (attempts.loc[attempts["is_correct"] == 0, "student_answer"].str.lower()
        != attempts.loc[attempts["is_correct"] == 0, "correct_answer"].str.lower()).all()
assert set(attempts["learning_gain"].unique()).issubset({0.012, 0.006, 0.002})

print(f"Simulated {len(attempts):,} answers across {WEEKS} weeks.")
display(attempts[["student_id", "week", "skill", "difficulty", "student_answer",
                  "is_correct", "learning_gain"]].head(8))
```

Cell 4 — Evaluate the adaptive tutor week by week

```python
overview = (
    weekly.groupby("week")
    .agg(
        accuracy=("weekly_accuracy", "mean"),
        avg_difficulty=("difficulty_this_week", "mean"),
        simulated_skill_state=("simulated_avg_skill_state", "mean"),
        students_leveling_up=("level_change", lambda s: int((s > 0).sum())),
        students_leveling_down=("level_change", lambda s: int((s < 0).sum()))
    )
    .reset_index()
)
display(overview.round(3))

plt.figure(figsize=(9, 4))
plt.plot(overview["week"], overview["accuracy"], marker="o", label="Answer accuracy")
plt.plot(overview["week"], overview["simulated_skill_state"], marker="o",
         label="Simulated skill state")
plt.ylim(0, 1)
plt.xticks(range(1, WEEKS + 1))
plt.xlabel("Week")
plt.ylabel("Score (0–1)")
plt.title("Weekly learning simulation (revised gain rule)")
plt.grid(alpha=0.2)
plt.legend()
plt.show()

plt.figure(figsize=(9, 4))
plt.plot(overview["week"], overview["avg_difficulty"], marker="o")
plt.xticks(range(1, WEEKS + 1))
plt.yticks([1, 2, 3], ["Easy", "Medium", "Hard"])
plt.ylim(0.8, 3.2)
plt.xlabel("Week")
plt.ylabel("Average assigned difficulty")
plt.title("Adaptive difficulty over time")
plt.grid(alpha=0.2)
plt.show()
```
